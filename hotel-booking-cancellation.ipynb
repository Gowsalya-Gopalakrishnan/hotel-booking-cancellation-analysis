```python
df['children'] = df['children'].fillna(df['children'].mode()[0])
```

```python
# cancellation % by hotel type
df.groupby('hotel')['is_canceled'].mean() * 100

# cancellation % by number of special requests
df.groupby('total_of_special_requests')['is_canceled'].mean() * 100

# cancellation % by market segment
df.groupby('market_segment')['is_canceled'].mean() * 100

# average lead time: canceled (1) vs not canceled (0)
df.groupby('is_canceled')['lead_time'].mean()
```

```python
sns.set_theme(style="whitegrid")
primary_color, secondary_color = "#2B5C8F", "#E05A47"

palette = [primary_color, secondary_color]

months_order = ["January", "February", "March", "April", "May", "June", 
                "July", "August", "September", "October", "November", "December"]
df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories=months_order, ordered=True)
```

```python
plt.figure(figsize=(6, 6)) #value_counts() how many time the unique values are appeared 

plt.pie(df["is_canceled"].value_counts(), labels = ["Not Cancaled", "Cancaled"], autopct='%1.1f%%', colors = palette, startangle=90, explode=(0, 0.05))
plt.title("Overall Booking Cancellation Ratio", fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(8, 5)) #bar chart countplot will helpus to show counts of the unique occurances and based on that will create the bar chart

sns.countplot(data = df, x = "hotel", hue = "is_canceled", palette = palette)
plt.title("Cancellation by Hotel Type", fontweight='bold')
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.legend(labels = ["Not Cancaled", "Cancaled"])
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data = df, x = "arrival_date_month", hue = "is_canceled", palette = palette)
plt.title("Cancellation by Month Date", fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.legend(labels = ["Not Canceled", "Canceled"])
plt.show()
```

```python
Top_10 = df[df["country"] != "Unknown"]["country"].value_counts().head(10).reset_index()

plt.figure(figsize=(9, 5))

sns.barplot(data = Top_10, x = "count", y = "country",  color = primary_color)
plt.title("Top 10 Guest Origin Countries", fontweight='bold')
plt.xlabel("Number of Bookings")
plt.ylabel("Country Code")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
monthly_adr = df.groupby(['arrival_date_month', 'hotel'])['adr'].mean().reset_index()
sns.lineplot(data=monthly_adr, x='arrival_date_month', y='adr', hue='hotel', marker='o', linewidth=2.5, palette=palette)
plt.xlabel("Month")
plt.ylabel("Average of ADR ($)")
plt.title("Average Daily Rate (ADR) by Month ($)", fontweight='bold')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.kdeplot(data=df, x='lead_time', hue='is_canceled', palette=palette, fill=True, common_norm=False)
plt.xlabel("Lead Time (Days)")
plt.ylabel("Density")
plt.title("Lead Time Distribution (Canceled vs Not Canceled)", fontweight='bold')
plt.legend(labels = ["Not Cancaled", "Cancaled"])
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data = df, x = "customer_type", hue = "is_canceled", palette = palette)
plt.xlabel("Customer Type")
plt.ylabel("Number of Bookings")
plt.title("Bookings & Cancellations by Customer Type", fontweight='bold')
plt.legend(labels = ["Not Cancaled", "Cancaled"])
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data = df, x = "market_segment", hue = "is_canceled", palette = palette, order = df["market_segment"].value_counts().index)
plt.xlabel("Market Segment")
plt.ylabel("Number of Bookings")
plt.title("Bookings & Cancellations by Market Segment", fontweight='bold')
plt.xticks(rotation=30)
plt.legend(labels=['Not Canceled', 'Canceled'])
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(data = df, x ="total_of_special_requests", y = "is_canceled", color = primary_color, errorbar=None)
plt.xlabel("Number of Special Requests")
plt.ylabel("Cancellation Rate")
plt.title("Cancellation Rate by Number of Special Requests", fontweight='bold')
plt.ylabel("Cancellation Rate")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='reserved_room_type', hue='is_canceled', palette=palette, order=sorted(df['reserved_room_type'].unique()))
plt.xlabel("Number of Special Requests")
plt.ylabel("Number of Bookings")
plt.title("Reserved Room Type Comparison", fontweight='bold')
plt.legend(labels=['Not Canceled', 'Canceled'])
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df , x='arrival_date_year', hue='hotel', palette=palette)
plt.xlabel("Year")
plt.ylabel("Number of Bookings")
plt.title("Number of Bookings Over Years (2015 - 2017)", fontweight='bold')
plt.show()
```
